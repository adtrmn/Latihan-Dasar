# Project - Tugas Praktek Day 1

Yang harus dilakukan program
  1. Minta anggaran bulanan sekali di awal.
  2. Minta nominal transaksi berulang kali sampai pengguna mengetik done.
  3. Setiap transaksi langsung dikelompokkan dan dicetak, misalnya Transaksi 1: Rp30,000 -> Kecil. Batasnya sama dengan Sesi 3: di bawah 100.000 Kecil, di bawah 1.000.000 Sedang, selebihnya Besar.
  4. Nominal 0 atau negatif ditolak: cetak Nominal harus lebih dari 0, lalu minta input lagi tanpa menghitungnya. Untuk lompat ke putaran berikutnya, pakai continue di dalam if, di tempat Anda biasa memakai break. continue melewati sisa baris di putaran itu, sedangkan break keluar dari loop.
  5. Setelah done, cetak ringkasan:
      - jumlah transaksi, total, dan rata-rata (format :,.1f)
      - transaksi terbesar dan terkecil
      - berapa transaksi Kecil, Sedang, dan Besar
      - sisa anggaran dan persentase terpakai (format :.1f)
      - status: terpakai sampai 80% Aman, di atas 80% sampai 100% Waspada, di atas 100% Melebihi anggaran
  6. Kalau tidak ada transaksi sama sekali, cetak Tidak ada transaksi. dan jangan hitung apa-apa. Rata-rata tidak boleh membagi dengan nol.


Aturan main
  1. Hanya pakai yang sudah kita pelajari. Belum boleh list, dictionary, atau fungsi (itu materi Sesi 5-8).
  2. Bagi kode jadi blok yang diberi komentar singkat (# input, # klasifikasi, # ringkasan).
  3. Tulis prediksi sebelum menjalankan tiap uji.
  4. Program boleh crash kalau pengguna mengetik huruf selain done (misalnya abc). Menanganinya adalah materi Sesi 11.

## Uji 1
Anggaran 5000000, transaksi 30000, 250000, 1200000, 75000, lalu done. Hitung dulu semua angka ringkasannya dengan tangan, baru jalankan dan bandingkan.

## Uji 2
Uji Batas - Anggaran 1000000, transaksi 99999, 100000, 999999, 1000000, lalu done. kelas tiap transaksi berurutan: Kecil, Sedang, Sedang, Besar

## Uji 3
Langsung done setelah anggaran: harus tercetak Tidak ada transaksi. dan tidak ada error.

## Uji 4
Anggaran 500000, transaksi 0, -5, 50000, lalu done: dua input pertama ditolak dengan pesan, ringkasan hanya menghitung satu transaksi.

## Worksheet

```python
# 1. membuat input anggaran bulanan
anggaran = int(input("Masukkan Anggaran Bulanan: "))

# 2. variabel awal
jumlah = 0
total = 0
kecil = 0
sedang = 0
besar = 0

# 3. Loop Transaksi
while True:
    teks = input("Transaksi: ")
    if teks == "done":
        break
    nominal = int(teks)
    if nominal <= 0:
        print("Nominal harus lebih dari 0")
        continue

     # TODO: klasifikasi Kecil/Sedang/Besar, cetak kelasnya, tambah penghitung
    elif nominal < 100000:
        print("Transaksi Kecil")
    elif nominal < 1000000:
        print("Transaksi Sedang")
    else:
        print("Transaksi Besar")
    jumlah += 1
    total += nominal

    # TODO: perbarui jumlah, total, terbesar, terkecil
    if nominal < 100000:
        kecil += 1
    elif nominal < 1000000:
        sedang += 1
    else:
        besar += 1
    if jumlah == 1: # memperbarui berdasarkan total list transaksi. 1 artinya dalah trx pertama, jadi bisa di jadikan terbesar dan terkecil
        terbesar = nominal
        terkecil = nominal
    else:
        if nominal > terbesar: # jika transaksi > terbesar, maka timpa terbesar
            terbesar = nominal
        if nominal < terkecil: # jika transaksi < terkecil, maka timpa terkecil
            terkecil = nominal

# 4. ringkasan
if jumlah == 0:
    print("Tidak ada transaksi.")
else:
    # TODO: semua perhitungan dan cetakan ringkasan di sini
    print(f"Total Transaksi: Rp. {total:,}")
    print(f"Rata-rata Transaksi: Rp.{total / jumlah:,}")
    print(f"Sisa Anggaran: Rp.{anggaran - total:,}")
    print(f"Terpakai {total / anggaran * 100:.1f}%")
    if total / anggaran * 100 > 100:
        print("Status: Melebihi Anggaran")
    print(f"Transaksi terkecil: Rp.{terkecil:,}")
    print(f"Transaksi terbesar: Rp.{terbesar:,}")
```
