## Sesi 3 - input() dan if/elif/else

```python
# Latihan 1 - prediksi

nominal = 100000

if nominal < 100000:
  print("Kecil")
elif nominal < 1000000:
  print("Sedang")
else:
  print("Besar")
```

```python
# Latihan 2 - prediksi
data = input("Masukkan angkanya: ")
nilai = int(data)

if nilai < 100000:
  print("Kecil")
elif nilai < 1000000:
  print("Sedang")
else:
  print("Besar")
```

```python
# Latihan 3 - cari bug

total = 0
while True:
    teks = input("Nominal: ")
    if teks == "done": # cek "done" SEBELUM int(), kalau tidak -> ValueError
        break          # keluar dari loop
    total += int(teks)
print(total) #untuk input dilakukan beberapa kali, tidak bisa dilakukan input langsung (30, 75, done) --> error
```
